```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.impute import SimpleImputer
```

```python
ecomshop = pd.read_csv("shop_smart_ecommerce.csv")
```

```python
ecomshop.head()
```

```python
ecomshop.shape
```

```python
ecomshop.isnull().any()
```

```python
print(ecomshop.dtypes)
```

```python
le = LabelEncoder()
ecomshop["Month"]= le.fit_transform(ecomshop["Month"])
ecomshop["VisitorType"]= le.fit_transform(ecomshop["VisitorType"])
```

```python
ecomshop.head()
```

```python
x= ecomshop.drop("Revenue",axis=1)
y= ecomshop[["Revenue"]]
```

```python
num_features = x.select_dtypes(include=["int64", "float64"]).columns
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
scaler = StandardScaler()
x_train_scale = scaler.fit_transform(x_train)
x_test_scale = scaler.transform(x_test)
```

```python
print(x_train_scale)

```

```python
model = DecisionTreeClassifier(max_depth= 6,
                               min_samples_split = 50,
                               min_samples_leaf=25,
                               random_state=42,
                               class_weight="balanced")
```

```python
model.fit(x_train_scale,y_train)
y_pred = model.predict(x_test_scale)
```

```python
from sklearn.metrics import f1_score,classification_report,confusion_matrix
print("f1Score:",f1_score(y_test,y_pred))
print("classification report:",classification_report(y_test,y_pred))
print("confusion_matrix:",confusion_matrix(y_test,y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV

model = DecisionTreeClassifier()

param_grid = {
    "max_depth": [4, 6, 8],
    "min_samples_leaf": [20, 30, 50]
}

grid = GridSearchCV(
    model,
    param_grid,
    scoring="f1",
    cv=5,
    
)

grid.fit(x_train_scale, y_train)

print("Best F1:", grid.best_score_)
print("Best params:", grid.best_params_)
```
